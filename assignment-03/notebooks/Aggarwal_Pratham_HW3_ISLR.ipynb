## ISLR: 3.7.4

### (a) Compare training RSS when the true relationship is linear.

**The cubic regression training RSS will be less than or equal to the linear regression training RSS.** The cubic model contains the linear model as the special case $\beta_2 = \beta_3 = 0$. Therefore, least squares can always achieve at least as small a training RSS with the cubic model. It will usually be lower, though equality is possible.

### (b) Compare test RSS when the true relationship is linear.

**We would generally expect the linear regression to have lower test RSS.** Since the true relationship is linear, the linear model is correctly specified. The cubic model can fit random fluctuations in the training data by estimating unnecessary higher-order terms, increasing variance and potentially worsening predictions on new observations. The exact test RSS on a particular sample can still vary.

### (c) Compare training RSS when the true relationship is nonlinear.

**The cubic regression training RSS will again be less than or equal to the linear regression training RSS.** This follows from the nested models: the cubic model can reproduce any linear fit by setting $\beta_2 = \beta_3 = 0$. The amount of nonlinearity does not change this training-RSS ordering.

### (d) Compare test RSS when the true relationship is nonlinear.

**There is not enough information to determine which model will have lower test RSS.** If the true relationship is only mildly nonlinear, the linear model may perform better because it has lower variance. If the nonlinearity is substantial and the cubic model captures it well, the cubic model may perform better by reducing approximation bias. The test-RSS comparison depends on the strength and form of the nonlinearity, as well as the noise and sample.
