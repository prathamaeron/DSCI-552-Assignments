## ISLR: 2.4.1

### (a) The sample size n is extremely large, and the number of predictors p is small.

**Better.** With very large $n$ and small $p$, a flexible method can estimate the relationship accurately and its increased variance is less problematic because there are many observations. Its lower bias is therefore likely to improve test performance.

### (b) The number of predictors p is extremely large, and the number of observations n is small.

**Worse.** When $p$ is very large relative to the small sample size, a flexible method has high variance and can overfit the training observations. The resulting variance is likely to outweigh the benefit of its lower bias.

### (c) The relationship between the predictors and response is highly non-linear.

**Better.** A highly non-linear relationship is poorly represented by an inflexible method, which would have high bias. A flexible method can capture the non-linearity and should therefore have lower test error, provided it is adequately regularized or tuned.

### (d) The variance of the error terms, i.e. $σ^2$ = Var(ε), is extremely high.

**Worse, or approximately the same.** If the error variance $\sigma^2$ is extremely high, irreducible noise dominates the prediction error. Increasing flexibility cannot reduce this noise and may add estimation variance, so it will generally not improve performance and may make it worse.

## ISLR: 2.4.7

### (a) Compute the Euclidean distance between each observation and the test point, X1 = X2 = X3 = 0.

The test point is $(0,0,0)$. Using $d=\sqrt{(X_1-0)^2+(X_2-0)^2+(X_3-0)^2}$, the distances are:

| Observation | Distance | Y |
| --- | ---: | --- |
| $(0,3,0)$ | $3$ | Red |
| $(2,0,0)$ | $2$ | Red |
| $(0,1,3)$ | $\sqrt{10}$ | Red |
| $(0,1,2)$ | $\sqrt{5}$ | Green |
| $(-1,0,1)$ | $\sqrt{2}$ | Green |
| $(1,1,1)$ | $\sqrt{3}$ | Red |

In increasing order, the distances are $\sqrt{2}, \sqrt{3}, 2, \sqrt{5}, 3, \sqrt{10}$.

### (b) What is our prediction with K = 1? Why?

With $K=1$, the prediction is **Green**. The closest observation is $(-1,0,1)$, whose distance is $\sqrt{2}$, and its response is Green.

### (c) What is our prediction with K = 3? Why?

With $K=3$, the prediction is **Red**. The three closest observations have responses Green, Red, and Red, so the majority vote is Red.

### (d) If the Bayes decision boundary in this problem is highly non-linear, then would we expect the best value for K to be large or small? Why?

We would expect the best $K$ to be **small**. A highly non-linear Bayes decision boundary requires a flexible classifier to capture local changes in the relationship between $X$ and $Y$. Small $K$ gives KNN greater flexibility, whereas large $K$ produces a smoother, more inflexible decision boundary.
