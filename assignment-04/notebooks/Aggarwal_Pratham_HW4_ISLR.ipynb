## ISLR: 4.8.3

For an observation \(X=x\), Bayes' rule assigns the class with the largest posterior probability. Since the denominator in Bayes' rule is the same for every class, this is equivalent to maximizing the class prior times the class-conditional density:

$$\hat{k}(x)=\arg\max_{k=1,\ldots,K}\;\pi_k f_k(x),$$

where \(\pi_k=P(Y=k)\) and, under the assumption in the question,

$$f_k(x)=\frac{1}{\sqrt{2\pi\sigma_k^2}}\exp\!\left(-\frac{(x-\mu_k)^2}{2\sigma_k^2}\right).$$

Taking the logarithm (which preserves the maximizing class) and dropping the class-independent constant \(-\tfrac12\log(2\pi)\), the discriminant for class \(k\) is

$$\delta_k(x)=\log\pi_k-\frac12\log\sigma_k^2-\frac{(x-\mu_k)^2}{2\sigma_k^2}.$$

Expanding the square gives

$$\delta_k(x)=-\frac{x^2}{2\sigma_k^2}+\frac{\mu_k}{\sigma_k^2}x+\left(\log\pi_k-\frac12\log\sigma_k^2-\frac{\mu_k^2}{2\sigma_k^2}\right).$$

Because the coefficient of \(x^2\) depends on the class-specific variance \(\sigma_k^2\), comparing two classes generally produces a quadratic equation in \(x\), not a linear one. Thus the Bayes classifier is **not generally linear; its discriminants and decision boundaries are quadratic**. If all class variances happen to be equal, the \(x^2\) terms cancel between classes and the boundary reduces to a linear one; this is the special-case assumption used by LDA, not the general QDA model.

## ISLR: 4.8.7

Let \(D\) denote issuing a dividend. The given class priors are \(P(D)=0.8\) and \(P(D^c)=0.2\). Both classes have variance \(36\), so their common standard deviation is \(6\). At \(X=4\), the normal densities are

$$f_{X\mid D}(4)=\frac{1}{6\sqrt{2\pi}}\exp\!\left(-\frac{(4-10)^2}{2(36)}\right)=\frac{1}{6\sqrt{2\pi}}e^{-1/2},$$

$$f_{X\mid D^c}(4)=\frac{1}{6\sqrt{2\pi}}\exp\!\left(-\frac{(4-0)^2}{2(36)}\right)=\frac{1}{6\sqrt{2\pi}}e^{-2/9}.$$

Using Bayes' theorem, the common normalizing factor cancels:

$$P(D\mid X=4)=\frac{0.8f_{X\mid D}(4)}{0.8f_{X\mid D}(4)+0.2f_{X\mid D^c}(4)}=\frac{0.8e^{-1/2}}{0.8e^{-1/2}+0.2e^{-2/9}}\approx 0.752.$$

**The predicted probability that the company issues a dividend is approximately 0.752 (75.2%).**
